```python
def caesar_cipher_encrypt(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            start = ord('A') if char.isupper() else ord('a')
            result += chr((ord(char) - start + shift) % 26 + start)
        else:
            result += char
    return result

def caesar_cipher_decrypt(text, shift):

    return caesar_cipher_encrypt(text, -shift)

message = "Hello, World! This is a confidential message."
shift_key = 3

print(f"Original Message: {message}")
print(f"Shift Key: {shift_key}")

encrypted_message = caesar_cipher_encrypt(message, shift_key)
print(f"Encrypted Message: {encrypted_message}")

decrypted_message = caesar_cipher_decrypt(encrypted_message, shift_key)
print(f"Decrypted Message: {decrypted_message}")
```
