```python
import socket

def resolve_domain_to_ip(domain):
    """
    Resolves a single domain name to its IPv4 address.

    Args:
        domain (str): The domain name to resolve (e.g., 'google.com').

    Returns:
        str: The IPv4 address of the domain, or 'Not Found' if resolution fails.
    """
    try:
        ip_address = socket.gethostbyname(domain)
        return ip_address
    except socket.gaierror:
        return "Not Found"

print("Function 'resolve_domain_to_ip' defined.")



domain_names = [
    "google.com",
    "microsoft.com",
    "openai.com",
    "amazon.com",
    "wikipedia.org",
    "this-domain-does-not-exist.xyz"
]

print("\n--- Resolving Domain Names ---")
for domain in domain_names:
    ip = resolve_domain_to_ip(domain)
    print(f"Domain: {domain:<30} IP Address: {ip}")

print("\nThis demonstrates how to quickly obtain IP addresses for a list of domains.")
```
